## 1. SparkSession 생성

환경변수 PYSPARK_PYTHON, PYSPARK_DRIVER_PYTHON을 현재 실행되고 있는 Python 인터프리터의 경로(`sys.executable`)로 설정한다.

```python
import os
import sys
os.environ['PYSPARK_PYTHON']=sys.executable
os.environ['PYSPARK_DRIVER_PYTHON']=sys.executable
```

SparkSession 객체를 생성하여 `spark` 변수에 저장한다.

```python
import pyspark
myConf=pyspark.SparkConf()
spark = pyspark.sql.SparkSession.builder\
    .master("local")\
    .appName("myApp")\
    .config(conf=myConf)\
    .getOrCreate()
```

## 2. Python 관련 환경변수 출력

사용하는 Python의 버전과 경로를 출력하고, 환경변수 PYSPARK_PYTHON, PYSPARK_DRIVER_PYTHON이 `sys.executable`과 일치하는지 확인한다.
